```python
# 1. Separate target variable (Loan_Status) from feature set
x = df.drop(columns=['Loan_Status'])   # x = all the input features (everything except the answer)
y = df['Loan_Status']                  # y = the answer we want to predict (0 = rejected, 1 = approved)

# Dependents was stored as nullable Int64 after extraction - cast to a plain numeric type
# (sklearn models expect plain numeric types, not pandas' nullable Int64)
x['Dependents'] = x['Dependents'].astype(int)
```

```python
## to split the data to train and test set
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn import svm
from sklearn.linear_model import LogisticRegression
```

```python
## Train Test Split 

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, stratify=y, random_state=10)
```

```python
x_train.shape, x_test.shape, df.shape
```

```python
#model = svm.SVC(kernel='rbf')
```

```python
model = LogisticRegression()
```

```python
model.fit(x_train, y_train)
```

```python
x_train_prediction = model.predict(x_train)
```

```python
## training data accuracy 
training_data_accuracy = accuracy_score(y_train,x_train_prediction)
```

```python
## print the accuracy score 
print(' Train data accuracy', training_data_accuracy)
```

```python
## test data 

x_test_prediction = model.predict(x_test)

## training data accuracy 
test_data_accuracy = accuracy_score(y_test, x_test_prediction)
## print the accuracy score 
print(' Test data accuracy', test_data_accuracy)
```
